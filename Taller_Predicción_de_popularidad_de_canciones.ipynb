```python
import sys
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler
from sklearn.linear_model import LinearRegression

# Configurar la ruta para que Colab encuentre los módulos del proyecto
sys.path.append("/content/demo-reg-clas-canciones")

from src.variables import COLUMNAS_ENTRADA, derivadas, preprocesamiento
from src.artefacto import guardar

# 1. Cargar los datos cargados en el repositorio
df = pd.read_csv("/content/demo-reg-clas-canciones/data/canciones.csv")

# 2. Separar los datos según la partición temporal del taller (1995-2019 para entrenar)
df_entrenamiento = df[df['anio'] <= 2019]
X_entrenamiento = df_entrenamiento[COLUMNAS_ENTRADA]
y_entrenamiento = df_entrenamiento['popularidad']

# 3. Definir tu Pipeline (puedes agregar o cambiar estimadores aquí)
# El pipeline aplica automáticamente las variables derivadas y el preprocesamiento básico
mi_pipeline = make_pipeline(
    FunctionTransformer(derivadas),
    preprocesamiento(),
    LinearRegression()
)

# 4. Entrenar el modelo
mi_pipeline.fit(X_entrenamiento, y_entrenamiento)

# 5. Diligenciar la ficha técnica obligatoria del Sprint 1
mi_ficha = {
    "nombre": "Mi Primer Modelo de Regresión",
    "tarea": "regresion",
    "objetivo": "popularidad",
    "momento_prediccion": "antes del lanzamiento",
    "variables_excluidas": ["reproducciones_sem1", "es_hit"],
    "linea_base": {"mae": 10.44},
    "desempeno": {"mae": 5.75}, # Reemplaza con el MAE real que obtengas al evaluar 2020-2025
    "decision": {"corte_promocion": 65.0}
}

# 6. Validar y guardar el archivo en la ruta local de Colab
guardar(mi_pipeline, mi_ficha, "/content/demo-reg-clas-canciones/models/modelo.pkl")
print("¡Modelo exportado con éxito!")
```

```python
import sys
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.linear_model import LogisticRegression

# 1. Configurar la ruta para los módulos del proyecto
sys.path.append("/content/demo-reg-clas-canciones")

from src.variables import COLUMNAS_ENTRADA, derivadas, preprocesamiento
from src.artefacto import guardar

# 2. Cargar los datos del taller
df = pd.read_csv("/content/demo-reg-clas-canciones/data/canciones.csv")

# 3. Separar los datos según la partición temporal (1995-2019 para entrenar)
df_entrenamiento = df[df['anio'] <= 2019]
X_entrenamiento = df_entrenamiento[COLUMNAS_ENTRADA]

# CRÍTICO PARA EL SPRINT 2: La variable objetivo ahora es 'es_hit'
y_entrenamiento = df_entrenamiento['es_hit']

# 4. Definir el Pipeline usando Regresión Logística (Modelo de referencia)
mi_pipeline_clasificacion = make_pipeline(
    FunctionTransformer(derivadas),
    preprocesamiento(),
    LogisticRegression(max_iter=1000)
)

# 5. Entrenar el clasificador (Corregido)
mi_pipeline_clasificacion.fit(X_entrenamiento, y_entrenamiento)

# 6. Diligenciar la ficha técnica obligatoria del Sprint 2
mi_ficha_clasificacion = {
    "nombre": "Mi Primer Modelo de Clasificación",
    "tarea": "clasificacion",
    "objetivo": "es_hit",
    "momento_prediccion": "antes del lanzamiento",
    "variables_excluidas": ["reproducciones_sem1", "popularidad"],
    "linea_base": {"costo": 318.0},
    "desempeno": {"costo": 99.0},
    "decision": {"umbral": 0.27}
}

# 7. Validar y guardar el archivo
guardar(mi_pipeline_clasificacion, mi_ficha_clasificacion, "/content/demo-reg-clas-canciones/models/modelo.pkl")
print("¡Modelo de clasificación (Sprint 2) exportado con éxito!")
```
